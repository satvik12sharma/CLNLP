**EXPERIMENT-1:DATA ANALYSIS**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read employee dataset
df = pd.read_csv("employee_information_100.csv")

# Display dataset
print(df)
```

Experiment-1.1

```python
avg_salary = df.groupby("Department")["Salary"].mean()

print("Average Salary of Each Department:")
print(avg_salary)

# Bar chart
avg_salary.plot(kind="bar")

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)
plt.show()
```

Experiment-1.2

```python
employee_count = df["Department"].value_counts()

print("Number of Employees in Each Department:")
print(employee_count)

employee_count.plot(kind="bar")

plt.title("Number of Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.show()
```

Experiment-1.3

```python
gender_count = df["Gender"].value_counts()

print("Gender Distribution:")
print(gender_count)

# Pie chart
plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%"
)

plt.title("Percentage of Male and Female Employees")
plt.show()
```

Experiment-1.4

```python
plt.hist(df["Salary"], bins=10, edgecolor="black")

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()
```

Experiment-1.5

```python
plt.scatter(df["Experience_Years"], df["Salary"])

plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()
```

Experiment-1.6

```python
top_10 = df.sort_values(by="Salary", ascending=False).head(10)

print("Top 10 Employees with Highest Salary:")
print(top_10[["Employee_ID", "Name", "Department", "Salary"]])
```

Experiment-1.7

```python
highest_salary = df.groupby("Department")["Salary"].max()

print("Highest Salary in Each Department:")
print(highest_salary)
```

Experiment-1.8

```python
average_salary = df["Salary"].mean()

print("Overall Average Salary:", average_salary)

above_average = df[df["Salary"] > average_salary]

print("\nEmployees earning more than the overall average salary:")
print(above_average[
    ["Employee_ID", "Name", "Department", "Salary"]
])
```

Experiment-1.9

```python
avg_experience = df.groupby("Department")["Experience_Years"].mean()

print("Average Experience in Each Department:")
print(avg_experience)

avg_experience.plot(kind="bar")

plt.title("Average Experience by Department")
plt.xlabel("Department")
plt.ylabel("Average Experience (Years)")
plt.xticks(rotation=0)
plt.show()
```

Experiment-1.10

```python
plt.hist(df["Age"], bins=10, edgecolor="black")

plt.title("Distribution of Employee Ages")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```
