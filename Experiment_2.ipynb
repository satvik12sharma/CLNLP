Experiment-2.1

```python
#import modules
import string
```

```python
#read text file
with open("2.1_text_data.txt","r") as file:
  text=file.read()
```

```python
#Upper case
countup=0
for i in text:
  if i.isupper():
    countup=countup+1
print("Number of Upper Cases=",countup)
cleaned_text=text.lower()
```

```python
#Punctuations
countpunc=0
punc=[]
for i in text:
  if i in string.punctuation:
    if i not in punc:
      punc.append(i)
      countpunc=countpunc+1
    cleaned_text=cleaned_text.replace(i,"")
print("Number of Types of Punctuations=",countpunc)
```

```python
#Numbers
countnum=0
for i in text:
  if i.isdigit():
    countnum=countnum+1
    cleaned_text=cleaned_text.replace(i,"")
print("Number of Numbers=",countnum)
```

```python
#Extra Whitespaces
counttotws=0
for i in text:
  if i.isspace():
    counttotws=counttotws+1
print("Number of Total White Spaces=",counttotws)
words=text.split()
reqdws=max(0,len(words)-1)
extrawhitespace=counttotws-reqdws
print("Number of Extra White Spaces=",extrawhitespace)
cleaned_text = ' '.join(cleaned_text.split())
```

```python
#Final Cleaned Text
print(cleaned_text)
```

```python
#Using Regular Expression
import re
```

```python
#Read Text File
with open("2.2_tokenization_data.txt","r") as file:
  text=file.read()
```

```python
# Sentence Tokenization
sentences = re.split(r'(?<=[.!?])\s+', text.strip())

# Word Tokenization
words = re.findall(r'\b\w+\b|[^\w\s]', text)
```

```python
# Display sentences
print("Sentence Tokenization:")
c=0
for sentence in sentences:
    c=c+1
    print(c, ":", sentence)

# Display words
print("Word Tokenization:")
c=0
for word in words:
    c=c+1
    print(c, ":", word)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
# Download required data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
#Read Text File
with open("2.3_clean_data.txt","r") as file:
  text=file.read()
```

```python
# Tokenize the text
tokens = word_tokenize(text)

# Get English stop words
stop_words = set(stopwords.words("english"))
```

```python
#Find Stop Words
found_stop_words = []
for i in words:
  if i.lower() in stop_words:
    found_stop_words.append(i)
```

```python
#Remove Stop Words
filtered_tokens = []
for i in tokens:
  if i.lower() not in stop_words:
    filtered_tokens.append(i)
```

```python
# Display original tokens
print("Original Tokens:")
print(tokens)

# Display stop words separately
print("Stop Words:")
print(found_stop_words)

# Display filtered tokens
print("Filtered Tokens:")
print(filtered_tokens)
```
