```python
#BPE Tokenization using Pretrained Model

from transformers import AutoTokenizer

# Load pretrained GPT-2 tokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")

# Read input data
with open("input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read().strip()

# Tokenize into subword units
tokens = tokenizer.tokenize(text)

# Convert tokens to token IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("\nGenerated Tokens:")
print(tokens)

print("\nCorresponding Token IDs:")
print(token_ids)

# Display token and ID together
print("\nToken - Token ID")
for token, token_id in zip(tokens, token_ids):
    print(f"{token:20} {token_id}")
```

```python
#BPE Tokenization without Pretrained Model

from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# Read input data
with open("input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read().strip()

print("Input Text:")
print(text)

# Create an empty BPE tokenizer
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))

# Split text into words before BPE processing
tokenizer.pre_tokenizer = Whitespace()

# Define BPE trainer
trainer = BpeTrainer(
    vocab_size=1000,
    min_frequency=1,
    special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]"]
)

# Train tokenizer on the input data
tokenizer.train(["input_sub_word_data.txt"], trainer)

# Encode the input text
encoding = tokenizer.encode(text)

# Get generated tokens and IDs
tokens = encoding.tokens
token_ids = encoding.ids

print("\nGenerated Tokens:")
print(tokens)

print("\nCorresponding Token IDs:")
print(token_ids)

# Display token and ID together
print("\nToken - Token ID")
for token, token_id in zip(tokens, token_ids):
    print(f"{token:20} {token_id}")
```

```python
#SentencePiece Tokenization using Pretrained Model

from transformers import T5Tokenizer

# Input file path
file_path = "input_sub_word_data.txt"

# Read input data
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read().strip()

print("Input Text:")
print(text)

# Load pretrained SentencePiece tokenizer
tokenizer = T5Tokenizer.from_pretrained("t5-small")

# Tokenize into subword units
tokens = tokenizer.tokenize(text)

# Convert tokens to IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("\nGenerated Tokens:")
print(tokens)

print("\nCorresponding Token IDs:")
print(token_ids)

# Display token and ID together
print("\nToken - Token ID")
for token, token_id in zip(tokens, token_ids):
    print(f"{token:20} {token_id}")
```

```python
# 5.2(b) SentencePiece Tokenization without Pretrained Model

import sentencepiece as spm

# Input file path
file_path = "input_sub_word_data.txt"

# Train a new SentencePiece model
spm.SentencePieceTrainer.train(
    input=file_path,
    model_prefix="my_sentencepiece",
    vocab_size=100,
    model_type="bpe",
    character_coverage=1.0,
    unk_id=0
)

# Load the newly trained model
sp = spm.SentencePieceProcessor(
    model_file="my_sentencepiece.model"
)

# Read input data
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read().strip()

print("Input Text:")
print(text)

# Generate subword tokens
tokens = sp.encode(text, out_type=str)

# Generate token IDs
token_ids = sp.encode(text, out_type=int)

print("\nGenerated Tokens:")
print(tokens)

print("\nCorresponding Token IDs:")
print(token_ids)

# Display token and ID together
print("\nToken - Token ID")
for token, token_id in zip(tokens, token_ids):
    print(f"{token:20} {token_id}")
```

Experiment-5.3

```python
#POS Tagging using spaCy

import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Given text
text = "The young student is reading an interesting book in the library."

# Process the text
doc = nlp(text)

print("Token\t\tPOS Tag\t\tDescription")
print("-" * 60)

# Display token, POS tag and description
for token in doc:
    if not token.is_punct:
        print(f"{token.text:15} {token.pos_:10} {spacy.explain(token.pos_)}")
```

```python
#POS Tagging using NLTK

import nltk
from nltk import word_tokenize, pos_tag

nltk.download("punkt")
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download("tagsets")

# Given sentence
text = "The young student is reading an interesting book in the library."

# Tokenize
tokens = word_tokenize(text)

# POS tagging
tagged_tokens = pos_tag(tokens)

# POS tag descriptions
pos_descriptions = {
    "DT": "Determiner",
    "JJ": "Adjective",
    "NN": "Noun, singular",
    "VBZ": "Verb, 3rd person singular present",
    "VBG": "Verb, gerund/present participle",
    "IN": "Preposition or subordinating conjunction"
}

print("Token\t\tPOS Tag\t\tDescription")
print("-" * 70)

for token, tag in tagged_tokens:
    if token.isalpha():
        description = pos_descriptions.get(tag, "Other")
        print(f"{token:15} {tag:10} {description}")
```

```python
import spacy
from collections import Counter

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Input file
file_path = "input_sub_word_data.txt"

# Read input data
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Process the text
doc = nlp(text)

# Count frequency of each token
token_frequency = Counter(
    token.text.lower()
    for token in doc
    if not token.is_space and not token.is_punct
)

print("Token\t\tPOS Tag\t\tDescription\t\tFrequency")
print("-" * 80)

# Store already displayed tokens
displayed_tokens = set()

# Display unique token information
for token in doc:
    token_lower = token.text.lower()

    # Ignore spaces and punctuation
    if token.is_space or token.is_punct:
        continue

    # Display each unique token only once
    if token_lower not in displayed_tokens:
        displayed_tokens.add(token_lower)

        print(
            f"{token.text:15}"
            f"{token.pos_:10}"
            f"{spacy.explain(token.pos_):30}"
            f"{token_frequency[token_lower]}"
        )
```

```python
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
import re

# Read corpus
file_path = "corpus.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Split into sentences
sentences = text.split("\n")

# Tokenize and convert to lowercase
tokenized_sentences = [
    re.findall(r'\b\w+\b', sentence.lower())
    for sentence in sentences
]

# Train Word2Vec model
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,
    window=2,
    min_count=1,
    workers=1,
    sg=0,          # CBOW
    epochs=100
)

# Function to calculate cosine similarity
def calculate_similarity(word1, word2):
    vector1 = model.wv[word1].reshape(1, -1)
    vector2 = model.wv[word2].reshape(1, -1)

    similarity = cosine_similarity(vector1, vector2)[0][0]

    return similarity


# Selected word pairs
word_pairs = [
    ("machine", "learning"),
    ("natural", "language"),
    ("deep", "learning")
]

# Display results
print("Cosine Similarity Results")
print("-" * 30)

for word1, word2 in word_pairs:
    similarity = calculate_similarity(word1, word2)

    print(f"{word1} - {word2}: {similarity:.4f}")
```
