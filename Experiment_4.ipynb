Experiment-4.1

```python
import nltk
import spacy
import csv
from collections import Counter
```

```python
nltk.download('punkt_tab')
```

```python
with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
words = nltk.word_tokenize(text)
```

```python
words = [word.lower() for word in words if word.isalpha()]
```

```python
frequency = Counter(words)
```

```python
with open("4.1_frequency.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Term", "Frequency"])
    for word, count in frequency.items():
        writer.writerow([word, count])
```

```python
print("Word frequencies saved to word_frequency.csv")
```

```python
print("\nTop 10 Most Frequent Terms:")

for word, count in frequency.most_common(10):
    print(word, ":", count)
```

```python
nlp = spacy.load("en_core_web_sm")

doc = nlp(text)

print("\nNamed Entities:")

for entity in doc.ents:
    print(entity.text, "->", entity.label_)
```

Experiment-4.2

```python
import re
import csv
from collections import Counter
```

```python
with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
text = text.lower()

words = re.findall(r'\b[a-zA-Z]+\b', text)

frequency = Counter(words)
```

```python
with open("4.2_frequency.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for term, count in frequency.items():
        writer.writerow([term, count])
```

```python
top_10 = frequency.most_common(10)

print("Top 10 Most Frequent Terms:")
print("---------------------------")

for term, count in top_10:
    print(term, ":", count)
```

Experiment-4.3

```python
import re
import math
from collections import Counter
```

```python
documents = [
    """Natural language processing is a field of
    artificial intelligence.""",

    """Natural language processing helps computers
    understand human language.""",

    """Machine learning is an important part of
    artificial intelligence."""
]
```

```python
tokenized_documents = []

for document in documents:

    # Convert to lowercase
    document = document.lower()

    # Remove punctuation
    document = re.sub(r'[^a-zA-Z\s]', '', document)

    # Tokenization
    tokens = document.split()

    tokenized_documents.append(tokens)
```

```python
print("TOKENIZED DOCUMENTS")

for i, tokens in enumerate(tokenized_documents):
    print("Document", i + 1, ":", tokens)
```

```python
term_frequencies = []

for tokens in tokenized_documents:

    word_count = Counter(tokens)
    total_words = len(tokens)

    tf = {}

    for term, count in word_count.items():
        tf[term] = count / total_words

    term_frequencies.append(tf)
```

```python
document_frequency = {}

all_terms = set()

for tokens in tokenized_documents:
    all_terms.update(tokens)

for term in all_terms:

    count = 0

    for tokens in tokenized_documents:
        if term in tokens:
            count += 1

    document_frequency[term] = count
```

```python
N = len(documents)

inverse_document_frequency = {}

for term in all_terms:

    df = document_frequency[term]

    inverse_document_frequency[term] = math.log(N / df)
```

```python
tf_idf = []

for tf in term_frequencies:

    document_tfidf = {}

    for term, value in tf.items():

        idf = inverse_document_frequency[term]

        document_tfidf[term] = value * idf

    tf_idf.append(document_tfidf)
```

```python
print("TF-IDF RESULTS")

for i in range(N):

    print("\nDocument", i + 1)
    print("-" * 65)

    print(f"{'Term':<15}{'TF':<12}{'DF':<8}"
          f"{'IDF':<12}{'TF-IDF':<12}")

    print("-" * 65)

    for term in sorted(term_frequencies[i]):

        tf = term_frequencies[i][term]
        df = document_frequency[term]
        idf = inverse_document_frequency[term]
        tfidf = tf_idf[i][term]

        print(f"{term:<15}"
              f"{tf:<12.4f}"
              f"{df:<8}"
              f"{idf:<12.4f}"
              f"{tfidf:<12.4f}")
```

```python
print("TOP 10 TERMS WITH HIGHEST TF-IDF")

for i in range(N):

    sorted_terms = sorted(
        tf_idf[i].items(),
        key=lambda x: x[1],
        reverse=True
    )

    print("\nDocument", i + 1)
    print("-" * 35)

    for term, score in sorted_terms[:10]:

        print(f"{term:<15} {score:.4f}")
```
