```python
# ============================================================
# 1. IMPORT LIBRARIES AND INITIALIZATION
# ============================================================

import os
import re
import json
import time
import tarfile
import random
import unicodedata
import warnings
import gc

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import torch

warnings.filterwarnings('ignore')

os.environ['HF_HUB_DISABLE_SYMLINKS_WARNING'] = '1'
os.environ['HF_HUB_DISABLE_XET'] = '1'

import huggingface_hub.file_download as fd

fd.are_symlinks_supported = lambda *a, **kw: False

from transformers import logging as hf_logging

hf_logging.set_verbosity_error()

# Set random seed and configuration

SEED = 42
N_SAMPLES = 200
LANG = 'hindi'

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device(
    'cuda' if torch.cuda.is_available() else 'cpu'
)

print(
    'Device:',
    device,
    '|',
    torch.cuda.get_device_name(0) if device.type == 'cuda' else 'CPU'
)


# ============================================================
# 2. DATASET: XL-SUM HINDI TEST SPLIT
# ============================================================

from huggingface_hub import hf_hub_download

# Download Hindi XL-Sum dataset archive

archive = hf_hub_download(
    'csebuetnlp/xlsum',
    f'data/{LANG}_XLSum_v2.0.tar.bz2',
    repo_type='dataset'
)

# Extract Hindi test dataset

with tarfile.open(archive) as tar:

    member = next(
        m for m in tar.getmembers()
        if m.name.endswith(f'{LANG}_test.jsonl')
    )

    test = pd.DataFrame([
        json.loads(line)
        for line in tar.extractfile(member).read().decode('utf-8').splitlines()
    ])

# Calculate article and summary word counts

test['text_words'] = test['text'].str.split().str.len()

test['summary_words'] = test['summary'].str.split().str.len()

print(
    f'XL-Sum {LANG.title()} test split: {len(test):,} articles'
)

print(
    test[['text_words', 'summary_words']]
    .describe()
    .loc[['mean', '50%', 'max']]
    .round(1)
)


# ============================================================
# 3. SELECT RANDOM EVALUATION SAMPLE
# ============================================================

sample = test.sample(
    N_SAMPLES,
    random_state=SEED
).reset_index(drop=True)

# Save selected sample

sample[
    ['id', 'url', 'title', 'summary', 'text']
].to_json(
    'xlsum_hindi_test_sample.jsonl',
    orient='records',
    lines=True,
    force_ascii=False
)

print(
    f'Evaluation sample: {len(sample)} articles'
)

print(
    'Median compression ratio:',
    f"{(sample['text_words'] / sample['summary_words']).median():.1f}x"
)

# Display sample article

ex = sample.iloc[0]

print('\nTITLE:', ex['title'])

print('SUMMARY:', ex['summary'])

print('ARTICLE:', ex['text'][:400], '...')


# ============================================================
# 4. EVALUATION METRICS: MULTILINGUAL ROUGE
# ============================================================

from rouge_score import rouge_scorer

# Custom Unicode tokenizer for Hindi text

class MultilingualTokenizer:

    def tokenize(self, text):

        text = ''.join(
            ' ' if unicodedata.category(ch)[0] in 'PS' else ch
            for ch in text.lower()
        )

        return text.split()


ROUGE_TYPES = [
    'rouge1',
    'rouge2',
    'rougeL'
]

# Default tokenizer

default_scorer = rouge_scorer.RougeScorer(
    ROUGE_TYPES,
    use_stemmer=True
)

# Multilingual tokenizer

ml_scorer = rouge_scorer.RougeScorer(
    ROUGE_TYPES,
    tokenizer=MultilingualTokenizer()
)

# Compare tokenization

ref = sample.loc[0, 'summary']

near_copy = sample.loc[0, 'title']

print(
    'Tokenized Hindi reference:',
    MultilingualTokenizer().tokenize(ref)[:12]
)

print(
    'Default ASCII tokenizer:',
    {
        k: round(v.fmeasure, 4)
        for k, v in default_scorer.score(ref, near_copy).items()
    }
)

print(
    'Multilingual tokenizer:',
    {
        k: round(v.fmeasure, 4)
        for k, v in ml_scorer.score(ref, near_copy).items()
    }
)


# ============================================================
# 5. SHARED GENERATION SETUP
# ============================================================

WHITESPACE = lambda t: re.sub(
    r'\s+',
    ' ',
    re.sub(r'\n+', ' ', t.strip())
)

# Common decoding parameters

GEN_KWARGS = dict(
    max_length=84,
    num_beams=4,
    no_repeat_ngram_size=2,
    early_stopping=True
)

# Prepare articles and reference summaries

articles = [
    WHITESPACE(t)
    for t in sample['text']
]

references = sample['summary'].tolist()

predictions = {}
timings = {}
n_params = {}


# ============================================================
# 6. mT5 MODEL: MULTILINGUAL TEXT-TO-TEXT TRANSFORMER
# ============================================================

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM
)

MT5_ID = 'csebuetnlp/mT5_multilingual_XLSum'

# Load pretrained mT5 tokenizer

mt5_tok = AutoTokenizer.from_pretrained(
    MT5_ID,
    revision='refs/pr/7',
    use_fast=False
)

# Load pretrained mT5 model

mt5 = AutoModelForSeq2SeqLM.from_pretrained(
    MT5_ID,
    revision='refs/pr/7',
    use_safetensors=True
).to(device)

# Calculate model parameters

n_params['mT5'] = sum(
    p.numel()
    for p in mt5.parameters()
)

print(
    f'mT5 parameters: {n_params["mT5"]/1e6:.1f}M'
)

print(
    f'Vocabulary: {mt5_tok.vocab_size:,}'
)

print(
    'mT5 subwords:',
    mt5_tok.tokenize(references[0])[:15]
)


# Function to generate summaries using mT5

@torch.no_grad()

def summarize_mt5(model, tokenizer, texts, batch_size=16):

    model.eval()

    outputs = []

    for i in range(0, len(texts), batch_size):

        enc = tokenizer(
            texts[i:i + batch_size],
            max_length=512,
            truncation=True,
            padding=True,
            return_tensors='pt'
        ).to(device)

        ids = model.generate(
            **enc,
            **GEN_KWARGS
        )

        outputs.extend(
            tokenizer.batch_decode(
                ids,
                skip_special_tokens=True,
                clean_up_tokenization_spaces=True
            )
        )

    return [
        o.strip()
        for o in outputs
    ]


# Generate mT5 summaries

t0 = time.time()

predictions['mT5'] = summarize_mt5(
    mt5,
    mt5_tok,
    articles
)

timings['mT5'] = time.time() - t0

print(
    f'Generated {len(predictions["mT5"])} mT5 summaries'
)

print(
    f'Inference time: {timings["mT5"]:.1f}s'
)

print(
    '\nmT5 sample summary:',
    predictions['mT5'][0]
)

# Release model memory

del mt5

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()


# ============================================================
# 7. mBART MODEL: MULTILINGUAL BIDIRECTIONAL AND AUTOREGRESSIVE
# ============================================================

MBART_ID = 'ai4bharat/IndicBART-XLSum'

# Load mBART tokenizer

mbart_tok = AutoTokenizer.from_pretrained(
    MBART_ID,
    do_lower_case=False,
    use_fast=False,
    keep_accents=True
)

# Load mBART model

mbart = AutoModelForSeq2SeqLM.from_pretrained(
    MBART_ID,
    revision='refs/pr/2',
    use_safetensors=True
).to(device)

# Calculate model parameters

n_params['mBART'] = sum(
    p.numel()
    for p in mbart.parameters()
)

# Special token IDs

bos_id = mbart_tok._convert_token_to_id_with_added_voc('<s>')

eos_id = mbart_tok._convert_token_to_id_with_added_voc('</s>')

pad_id = mbart_tok._convert_token_to_id_with_added_voc('<pad>')

hi_id = mbart_tok._convert_token_to_id_with_added_voc('<2hi>')

print(
    f'mBART parameters: {n_params["mBART"]/1e6:.1f}M'
)

print(
    f'Vocabulary: {len(mbart_tok):,}'
)

print(
    'Hindi token ID:',
    hi_id
)

print(
    'mBART subwords:',
    mbart_tok.tokenize(references[0])[:15]
)


# Function to generate summaries using mBART

@torch.no_grad()

def summarize_mbart(model, tokenizer, texts, batch_size=16):

    model.eval()

    outputs = []

    for i in range(0, len(texts), batch_size):

        # Preserve language token during truncation

        batch_inputs = [
            tokenizer.convert_tokens_to_string(
                tokenizer.tokenize(t)[:509]
            ) + ' </s> <2hi>'
            for t in texts[i:i + batch_size]
        ]

        enc = tokenizer(
            batch_inputs,
            padding=True,
            add_special_tokens=False,
            return_token_type_ids=False,
            return_tensors='pt'
        ).to(device)

        ids = model.generate(
            **enc,
            **GEN_KWARGS,
            decoder_start_token_id=hi_id,
            bos_token_id=bos_id,
            eos_token_id=eos_id,
            pad_token_id=pad_id
        )

        outputs.extend(
            tokenizer.batch_decode(
                ids,
                skip_special_tokens=True,
                clean_up_tokenization_spaces=False
            )
        )

    return [
        o.strip()
        for o in outputs
    ]


# Generate mBART summaries

t0 = time.time()

predictions['mBART'] = summarize_mbart(
    mbart,
    mbart_tok,
    articles
)

timings['mBART'] = time.time() - t0

print(
    f'Generated {len(predictions["mBART"])} mBART summaries'
)

print(
    f'Inference time: {timings["mBART"]:.1f}s'
)

print(
    '\nmBART sample summary:',
    predictions['mBART'][0]
)

# Release model memory

del mbart

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()


# ============================================================
# 8. COMPARATIVE EVALUATION: ROUGE AND BERTSCORE
# ============================================================

from bert_score import score as bert_score

per_sample = {}

for name, preds in predictions.items():

    # Calculate ROUGE scores

    rows = [
        ml_scorer.score(r, p)
        for r, p in zip(references, preds)
    ]

    # Calculate multilingual BERTScore

    P, R, F = bert_score(
        preds,
        references,
        model_type='google-bert/bert-base-multilingual-cased',
        num_layers=9,
        batch_size=32,
        device=str(device),
        verbose=False
    )

    per_sample[name] = pd.DataFrame({

        'ROUGE-1': [
            s['rouge1'].fmeasure
            for s in rows
        ],

        'ROUGE-2': [
            s['rouge2'].fmeasure
            for s in rows
        ],

        'ROUGE-L': [
            s['rougeL'].fmeasure
            for s in rows
        ],

        'BERTScore-P': P.cpu().numpy(),

        'BERTScore-R': R.cpu().numpy(),

        'BERTScore-F1': F.cpu().numpy(),

        'Gen length (words)': [
            len(p.split())
            for p in preds
        ]

    })


# Create comparative summary

summary = pd.DataFrame({
    k: v.mean()
    for k, v in per_sample.items()
}).T

# Convert evaluation scores into percentages

for col in [
    'ROUGE-1',
    'ROUGE-2',
    'ROUGE-L',
    'BERTScore-P',
    'BERTScore-R',
    'BERTScore-F1'
]:

    summary[col] *= 100

# Add additional information

summary['Reference length (words)'] = (
    sample['summary_words'].mean()
)

summary['Inference time (s)'] = pd.Series(timings)

summary['Params (M)'] = (
    pd.Series(n_params) / 1e6
)

print('\nFINAL MODEL EVALUATION')

print(summary.round(2))


# ============================================================
# 9. PAIRED COMPARISON AND BOOTSTRAP CONFIDENCE INTERVAL
# ============================================================

rng = np.random.default_rng(SEED)

print(
    f'{"Metric":<14} | {"mT5 wins":>8} | '
    f'{"mBART wins":>10} | {"Ties":>6} | '
    f'{"Difference":>16} | 95% Bootstrap CI'
)

print('-' * 95)

for m in [
    'ROUGE-1',
    'ROUGE-2',
    'ROUGE-L',
    'BERTScore-F1'
]:

    diff = (
        per_sample['mBART'][m]
        - per_sample['mT5'][m]
    ).values * 100

    # Generate 2000 bootstrap samples

    boots = [
        rng.choice(diff, len(diff)).mean()
        for _ in range(2000)
    ]

    lo, hi = np.percentile(
        boots,
        [2.5, 97.5]
    )

    print(
        f'{m:<14} | '
        f'{(diff < -1e-6).sum():>8} | '
        f'{(diff > 1e-6).sum():>10} | '
        f'{(np.abs(diff) <= 1e-6).sum():>6} | '
        f'{diff.mean():>+16.2f} | '
        f'[{lo:+.2f}, {hi:+.2f}]'
    )


# ============================================================
# 10. PERFORMANCE COMPARISON VISUALIZATION
# ============================================================

metrics = [
    'ROUGE-1',
    'ROUGE-2',
    'ROUGE-L',
    'BERTScore-F1'
]

fig, ax = plt.subplots(
    1,
    2,
    figsize=(14, 4.5)
)

# Bar chart

summary[metrics].T.plot.bar(
    ax=ax[0],
    rot=0,
    color=['#4c72b0', '#dd8452'],
    width=0.75
)

ax[0].set(
    title=f'Mean Evaluation Scores on {N_SAMPLES} XL-Sum Hindi Test Articles',
    ylabel='Score (x100)'
)

for c in ax[0].containers:

    ax[0].bar_label(
        c,
        fmt='%.2f',
        fontsize=8,
        padding=2
    )


# Prepare data for boxplot

long = pd.concat([
    d[metrics].assign(Model=k)
    for k, d in per_sample.items()
]).melt(
    id_vars='Model',
    var_name='Metric'
)

# Boxplot

sns.boxplot(
    data=long,
    x='Metric',
    y='value',
    hue='Model',
    ax=ax[1],
    palette=['#4c72b0', '#dd8452'],
    fliersize=1.5
)

ax[1].set(
    title='Per-Article Score Distributions (ROUGE and mBERT)',
    ylabel='F1 Score'
)

plt.tight_layout()

plt.savefig(
    '8_performance_comparison.png',
    dpi=300
)

plt.show()


# ============================================================
# 11. QUALITATIVE COMPARISON AND CSV EXPORT
# ============================================================

out = sample[
    ['id', 'title', 'summary']
].rename(
    columns={'summary': 'reference'}
)

# Add generated summaries and evaluation scores

for name in predictions:

    out[f'{name}_summary'] = predictions[name]

    out[f'{name}_rougeL'] = (
        per_sample[name]['ROUGE-L'] * 100
    ).round(2)

    out[f'{name}_mBERT_F1'] = (
        per_sample[name]['BERTScore-F1'] * 100
    ).round(2)


# Export all generated summaries

out.to_csv(
    '8_generated_summaries.csv',
    index=False,
    encoding='utf-8-sig'
)

print(
    'All 200 generated Hindi summaries saved successfully!'
)


# Display first three comparisons

for idx in range(3):

    print(
        f'\n=== Sample {idx+1}: {out.loc[idx, "title"]} ==='
    )

    print(
        f'Reference: {out.loc[idx, "reference"]}'
    )

    print(
        f'mT5: {out.loc[idx, "mT5_summary"]}'
    )

    print(
        f'mT5 ROUGE-L: {out.loc[idx, "mT5_rougeL"]}'
    )

    print(
        f'mT5 BERTScore: {out.loc[idx, "mT5_mBERT_F1"]}'
    )

    print(
        f'mBART: {out.loc[idx, "mBART_summary"]}'
    )

    print(
        f'mBART ROUGE-L: {out.loc[idx, "mBART_rougeL"]}'
    )

    print(
        f'mBART BERTScore: {out.loc[idx, "mBART_mBERT_F1"]}'
    )


# ============================================================
# 12. FINAL MODEL COMPARISON
# ============================================================

print('\nFINAL MODEL COMPARISON')

print(summary.round(2))

# Identify best model using BERTScore F1

best_model = summary['BERTScore-F1'].idxmax()

print(
    '\nBest Performing Model:',
    best_model
)

print(
    'Best BERTScore F1:',
    round(summary.loc[best_model, 'BERTScore-F1'], 2)
)

# Save final evaluation results

summary.round(2).to_csv(
    '8_final_evaluation_results.csv'
)

print('\nEXPERIMENT 8 COMPLETED SUCCESSFULLY!')

print('\nGenerated Files:')

print('1. xlsum_hindi_test_sample.jsonl')

print('2. 8_generated_summaries.csv')

print('3. 8_performance_comparison.png')

print('4. 8_final_evaluation_results.csv')

```
