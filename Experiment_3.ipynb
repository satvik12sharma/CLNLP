**EXPERIMENT-3:STEMMING, LEMMATIZATION AND REGULAR EXPRESSION**

Experiment-3.1

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connected",
    "connection",
    "computers"
]

print("Original Word\tStem")
print("------------------------")

for word in words:
    stem = stemmer.stem(word)
    print(f"{word}\t\t{stem}")
```

Experiment-3.2

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

words = [
    ("cats", "n"),
    ("dogs", "n"),
    ("running", "v"),
    ("runs", "v"),
    ("ran", "v"),
    ("studies", "v"),
    ("studying", "v"),
    ("better", "a"),
    ("children", "n"),
    ("mice", "n"),
    ("went", "v"),
    ("ate", "v"),
    ("leaves", "n"),
    ("caring", "v")
]

print("Original Word\tLemma")
print("-------------------------")

for word, pos in words:
    lemma = lemmatizer.lemmatize(word, pos)
    print(f"{word}\t\t{lemma}")
```

```python
# Experiment 3.3: Information Extraction using Regular Expressions

import re

text = """
Welcome to the Natural Language Processing workshop! For more information,
visit https://www.nlpworkshop.com or www.python.org. You can contact the
coordinator at nlpworkshop@gmail.com or support@python.org. For registration,
call +91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop
and @PythonLearner. Share your experience using #NLP, #Python, and
#MachineLearning. You can also visit https://github.com/NLPWorkshop
for the latest updates.
"""

# 1. Extract Email Addresses
emails = re.findall(
    r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}',
    text
)

# 2. Extract URLs
urls = re.findall(
    r'https?://[^\s]+|www\.[^\s]+',
    text
)

# 3. Extract Mobile Numbers
mobile_numbers = re.findall(
    r'(?:\+91[-\s]?)?[6-9]\d{9}',
    text
)

# 4. Extract Hashtags
hashtags = re.findall(
    r'#[A-Za-z0-9_]+',
    text
)

# 5. Extract Mentions
mentions = re.findall(
    r'@[A-Za-z0-9_]+',
    text
)

print("Email Addresses:")
print(emails)

print("\nURLs:")
print(urls)

print("\nMobile Numbers:")
print(mobile_numbers)

print("\nHashtags:")
print(hashtags)

print("\nMentions:")
print(mentions)
```
